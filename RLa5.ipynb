```python
import gymnasium as gym


def run_random_agent(environment_name):

    # Create the selected environment
    env = gym.make(environment_name)

    # Reset the environment
    state, info = env.reset()

    # Initialize total reward and step count
    total_reward = 0
    steps = 0

    print("\n--------------------------------")
    print("Environment:", environment_name)
    print("Initial State:", state)
    print("Action Space:", env.action_space)

    # Continue until the episode terminates
    while True:

        # Select an action randomly
        action = env.action_space.sample()

        # Execute the action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Display step information
        print("Step:", steps + 1)
        print("Action:", action)
        print("Next State:", next_state)
        print("Reward:", reward)

        # Calculate total reward
        total_reward += reward

        # Increase step count
        steps += 1

        # Update state
        state = next_state

        # Stop when episode terminates
        if terminated or truncated:
            break

    # Calculate average reward
    average_reward = total_reward / steps

    print("\nEpisode Terminated")
    print("Total Steps:", steps)
    print("Total Reward:", total_reward)
    print("Average Reward:", average_reward)

    env.close()


run_random_agent("CartPole-v1")
run_random_agent("MountainCar-v0")
run_random_agent("FrozenLake-v1")
```
