```python
# Greedy Multi-Armed Bandit using Arithmetic Mean

num_machines = int(input("Enter number of machines: "))

machines = {}

for i in range(num_machines):
    name = input(f"\nEnter name of machine {i + 1}: ")
    rewards = list(map(float, input("Enter observed rewards separated by spaces: ").split()))
    machines[name] = rewards

print("\n--- Estimated Q(a) Values ---")

q_values = {}

for machine, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[machine] = q

    # Display integers without .0
    display_rewards = [int(r) if r.is_integer() else r for r in rewards]

    print(f"\nMachine {machine}")
    print(f"Rewards: {display_rewards}")
    print(f"Q({machine}) = {q:g}")

# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```

```python
# Greedy Multi-Armed Bandit using Incremental Update

num_machines = int(input("Enter number of machines: "))

machines = {}
q_values = {}
counts = {}

# Initialize machines
for i in range(num_machines):
    name = input(f"\nEnter name of machine {i + 1}: ")
    rewards = list(map(float, input("Enter observed rewards separated by spaces: ").split()))

    machines[name] = rewards
    q_values[name] = 0.0
    counts[name] = 0

print("\n--- Incremental Q(a) Updates ---")

# Process every observed reward
for machine, rewards in machines.items():

    print(f"\nMachine {machine}")

    for reward in rewards:
        # Increment number of plays
        counts[machine] += 1

        # Incremental update formula
        q_values[machine] = (
            q_values[machine]
            + (reward - q_values[machine]) / counts[machine]
        )

        print(
            f"Reward = {reward:g}, "
            f"N({machine}) = {counts[machine]}, "
            f"Q({machine}) = {q_values[machine]:.2f}"
        )

        # Identify current greedy machine
        selected_machine = max(q_values, key=q_values.get)

        print(f"Current Greedy Selection: {selected_machine}")

# Final results
print("\n--- Final Q(a) Values ---")

for machine in machines:
    print(f"Q({machine}) = {q_values[machine]:.2f}")

# Final greedy selection
selected_machine = max(q_values, key=q_values.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```
